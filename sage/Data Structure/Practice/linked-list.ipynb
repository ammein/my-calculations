# Linked List
A linked list is a linear data structure where elements are stored in nodes. Each node contains:

- A **value** (the data it holds).
- A **reference** (or pointer) to the next node in the sequence.

This structure forms a chain of nodes, starting from a **head** (the first node) and ending at a node whose **next** reference is null. Linked lists are dynamic, meaning they can grow or shrink as needed, unlike arrays with fixed sizes.

```python
from sage.all import *
from sage.graphs.graph_generators import graphs
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

## 1. Create a Linked List
Code defines a LinkedList class and a Node class to represent individual elements.

```python

class LinkedList:
    def __init__(self):
        self.head = None
        # For SageMath only
        self.values = []

    def append(self, value):
        node = Node(value)

        # For SageMath
        self.values.append(value)

        # If head is empty, assign head to new node and return immediately
        if self.head is None:
            self.head = node
            return

        # if head is not empty, set to new variable for current head
        current = self.head

        # if current.next is available, assign to current head variable onto next to traverse
        while current.next:
            current = current.next

        # Assign to next node to continue traverse
        current.next = node

    def printList(self):
        current = self.head
        result = ""

        # When current is available, assign to result from current.value and traverse back from current.next
        while current:
            result += str(current.value)+"->"
            current = current.next
        print(result + "null")

list = LinkedList()
list.append("A")
list.append("B")
list.printList()

# SageMath graph
g = graphs.PathGraph(Integer(len(list.values) + 1))
label = list.values
label.append("null")
g.plot(vertex_size=500, vertex_labels=dict(zip(g, label)))
```
