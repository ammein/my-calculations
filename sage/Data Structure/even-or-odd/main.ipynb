```python
# A simple Python3 program to find
# maximum score that
# maximizing player can get
import math
from sage.graphs.digraph import DiGraph
from sage.rings.integer import Integer

D = DiGraph({}, multiedges=True)

# a % b = a - floor(a / b) * b
def even_or_odd(n):
    remain = Integer(n) % Integer(2)
    calc = Integer(n) - Integer(2)
    temp = Integer(n)
    while calc != remain:
        D.add_edge(temp, calc, '{}'.format(calc))
        temp = calc
        calc -= Integer(2)
    # Finalize after while loop
    D.add_edge(temp, calc, '{}'.format(calc))
# D.set_edges(edge_color='white')
D.graphplot(vertex_size=Integer(100), edge_labels=True,
                  color_by_label=True, edge_style='dashed')
even_or_odd(5)
D.plot()
```
